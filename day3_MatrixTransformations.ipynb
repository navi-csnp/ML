## Matrix Transformations

**Vectors transforms the space itself**

**Matric moves vectors**

v_new = A @ v

A matrix can: Stretch, Shrink, Rotate, Flip, Skew (Shear)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# SCALING
# X direction stretched, y direction unchanged 

A = np.array([[2, 1], [0, 1]])
v = np.array([1, 1])
print(A @ v)
```

```python
# ROTATION
# R(θ) = [[cos(θ) -sin(θ)], 
#        [sin(θ) cos(θ)]]

theta = np.pi / 2 #90 degrees

R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])
v = np.array([1, 0])
r = np.round(R @ v, 2)

print(r)
```

```python
# ROTATION PLOT
fig, axes = plt.subplots()
axes.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original vector')
axes.quiver(0, 0, r[0], r[1], angles='xy', scale_units='xy', scale=1, color='b', label='After rotation')

axes.set_xlim(-2, 2)
axes.set_ylim(-2, 2)


axes.set_title('Rotation and normal matrix')

axes.grid()
axes.set_aspect('equal')
axes.legend()

plt.show()
```

```python
v = np.array([2, 1])
theta = np.pi / 4 #45 degrees
r = np.round(R @ v, 2)
print(r)
```

```python
# ROTATION PLOT
fig, axes = plt.subplots()
axes.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original vector')
axes.quiver(0, 0, r[0], r[1], angles='xy', scale_units='xy', scale=1, color='b', label='After rotation')

axes.set_xlim(-3, 3)
axes.set_ylim(-3, 3)


axes.set_title('Rotation and normal matrix')
print(np.linalg.norm(v), np.linalg.norm(r))

axes.grid()
axes.set_aspect('equal')
axes.legend()

plt.show()
```

```python
#SHEAR

S = np.array([[1, 1], [0, 1]])
x = [1, 0]
y = [0, 2]

s_x = S @ x
s_y = S @ y

print(s_x, s_y) # s_y will be stretched
```

```python
# SHEAR PLOT - X axis is not changed in this example, only y is changed
fig, axes = plt.subplots()
axes.quiver(0, 0, y[0], y[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original vector')
axes.quiver(0, 0, s_y[0], s_y[1], angles='xy', scale_units='xy', scale=1, color='b', label='After Shear')

axes.set_xlim(-3, 3)
axes.set_ylim(-3, 3)


axes.set_title('Rotation and normal matrix')
print(np.linalg.norm(v), np.linalg.norm(r))

axes.grid()
axes.set_aspect('equal')
axes.legend()

plt.show()
```
