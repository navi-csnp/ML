```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#2D vector - PLotting a Single Vector
u = 3
v = 2

#Define the origin
origin_x = 0
origin_y = 0

#Create the plot
fig, axes = plt.subplots()
axes.quiver(origin_x, origin_y, u, v, angles='xy', scale_units='xy', scale=1, color='r')
axes.set_xlim(0,5)
axes.set_ylim(0,5)

axes.set_xticks(np.arange(0, 6, 2))
axes.set_yticks(np.arange(0, 6, 1))

plt.show()
```

```python
#Plotting multiple vectors

V = np.array([[1, 1], [-2, 2], [4, -7]])
origin_x, origin_y = np.zeros_like(V.T)

u, v = V.T
print(u, v)

fig, axes = plt.subplots()
axes.quiver(origin_x, origin_y, u, v, angles='xy', scale_units='xy', scale=1, color=['r','b','g'])

#Setting axis limits
max_val = np.amax(abs(V)) * 1.1 #(1.1 for some padding)
axes.set_xlim([-max_val, max_val])
axes.set_ylim([-max_val, max_val])
axes.set_xlabel('X-Axis')
axes.set_ylabel('Y-Axis')
axes.set_title('Multiple vectors from the origin')

axes.grid()
axes.set_aspect('equal')

plt.show()
```

```python
#Plot individual vectors and sum of the vectors and visualize

a = np.array([2, 3])
b = np.array([4, -1])

c = a + b

origin_x, origin_y = [0, 0, 0], [0, 0, 0]

u = [a[0], b[0], c[0]]
v = [a[1], b[1], c[1]]

fig, axes = plt.subplots()
axes.quiver(origin_x, origin_y, u, v, angles='xy', scale_units='xy', scale=1, color=['r','g','b'])

axes.set_xlim([-8, 8])
axes.set_ylim([-5, 5])

axes.set_xlabel('X axis')
axes.set_ylabel('Y axis')
axes.set_title('Plot individual vectors, sum of the vectors and visualize')

axes.grid()
axes.set_aspect('equal')

plt.show()
```
