## Dot Product visualization

a.b = ||a|| ||b|| cos Θ

Dot product depends on the angle between vectors.

The dot product changes based on angle:
- Smaller angle → larger value
- 90° → zero
- Greater than 90° → negative

This shows dot product measures similarity.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([2, 3])
b = np.array([1, 2])

origin_x = np.zeros_like(a.T)
origin_y = np.zeros_like(a.T)
```

```python
# Plotting vector 'a'

u = a[0]
v = a[1]

fig, axes = plt.subplots()
axes.quiver(origin_x, origin_y, u, v, angles='xy', scale_units='xy', scale=1, color=['r'])

axes.set_xlim([0, 5])
axes.set_ylim([0, 5])

axes.set_xlabel('X axis')
axes.set_ylabel('Y axis')
axes.set_title('Plotting vector "A"')

axes.grid()
axes.set_aspect('equal')

plt.show()
```

```python
# Plotting vector 'b'

u = b[0]
v = b[1]

fig, axes = plt.subplots()
axes.quiver(origin_x, origin_y, u, v, angles='xy', scale_units='xy', scale=1, color=['r'])

axes.set_xlim([0, 5])
axes.set_ylim([0, 5])

axes.set_xlabel('X axis')
axes.set_ylabel('Y axis')
axes.set_title('Plotting vector "B"')

axes.grid()
axes.set_aspect('equal')

plt.show()
```

```python
dot = np.dot(a, b)

fig, axes = plt.subplots()

axes.quiver(0, 0, a[0], a[1],angles='xy', scale_units='xy', scale=1,color='r', label='Vector a')
axes.quiver(0, 0, b[0], b[1],angles='xy', scale_units='xy', scale=1,color='b', label='Vector b')

axes.set_xlim(0, 5)
axes.set_ylim(0, 5)
axes.set_xticks(np.arange(0, 5, 1))
axes.set_yticks(np.arange(0, 5, 1))

axes.set_title(f"Dot Product = {dot}")

axes.grid()
axes.set_aspect('equal')
axes.legend()

plt.show()
```

```python
vectors_b = [
    np.array([1, 2]),     # similar
    np.array([1, -2]),    # angled
    np.array([-1, -2]),   # opposite
    np.array([4, 0])      # perpendicular
]

for b in vectors_b:
    dot = np.dot(a, b)

    fig, axes = plt.subplots()
    
    axes.quiver(0, 0, a[0], a[1],angles='xy', scale_units='xy', scale=1,color='r', label='Vector a')
    axes.quiver(0, 0, b[0], b[1],angles='xy', scale_units='xy', scale=1,color='b', label='Vector b')
    
    axes.set_xlim(-5, 5)
    axes.set_ylim(-5, 5)
    axes.set_xticks(np.arange(-5, 5, 1))
    axes.set_yticks(np.arange(-5, 5, 1))
    
    axes.set_title(f"Dot Product = {dot}")
    
    axes.grid()
    axes.set_aspect('equal')
    axes.legend()
    
    plt.show()
```

## Angle vs Dot Product

As angle increases:
- Dot product decreases
- At 90°, it becomes zero
- After 90°, it becomes negative

This shows dot product depends on cosine of angle.

```python
'''
b = cos(θ).a + sin(θ).p

where 
a = direction of vector a (forward direction)
p = perpendicular direction (sideways direction - 90 deg turn)
'''

a_unit = a / np.linalg.norm(a)

perp = np.array([-a[1], a[0]])
perp_unit = perp / np.linalg.norm(perp)

angles = [0, 30, 60, 90, 120, 150, 180]

for theta_deg in angles:
    theta = np.deg2rad(theta_deg)

    # construct b relative to a
    b = np.cos(theta) * a_unit + np.sin(theta) * perp_unit
    dot = np.dot(a, b)

    fig, axes = plt.subplots()
    
    axes.quiver(0, 0, a[0], a[1],angles='xy', scale_units='xy', scale=1, color='r', label='Vector a')
    axes.quiver(0, 0, b[0], b[1],angles='xy', scale_units='xy', scale=1, color='b', label='Vector b')
    
    axes.set_xlim(-5, 5)
    axes.set_ylim(-5, 5)
    axes.set_xticks(np.arange(-5, 5, 1))
    axes.set_yticks(np.arange(-5, 5, 1))
    
    axes.set_title(f"Dot Product = {dot}" + f", Angle = {theta_deg}")
    
    axes.grid()
    axes.set_aspect('equal')
    axes.legend()
    
    plt.show()
```
